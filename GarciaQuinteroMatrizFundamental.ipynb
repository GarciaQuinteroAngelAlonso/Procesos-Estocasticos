# Análisis de Simulación: Serpientes y Escaleras


# 1. Introducción

Este reporte presenta el análisis del número promedio de movimientos necesarios para completar un juego de serpientes y escaleras en un tablero de 20 casillas.

El problema se estudia mediante técnicas de simulación estocástica y cadenas de Markov, permitiendo aproximar el valor esperado del número de tiradas requeridas para alcanzar la meta.


# 2. Definición del Escenario

El tablero consiste en un recorrido discreto de 20 casillas.

La posición del jugador en el turno $t$ se representa mediante:

$$
P_t
$$

donde:

$$
P_t \in \{0,1,2,\dots,20\}
$$

La casilla $0$ representa el inicio y la casilla $20$ corresponde a la meta.

## 2.1 Reglas de transición

En cada turno se lanza un dado justo de seis caras.

El movimiento base del jugador está dado por:

$$
P_{t+1}=P_t+D
$$

donde:

$$
D \in \{1,2,3,4,5,6\}
$$

y cada resultado ocurre con probabilidad:

$$
P(D=k)=\frac{1}{6}
$$

para:

$$
k=1,2,3,4,5,6
$$


## 2.2 Escaleras y serpientes

El tablero contiene transiciones especiales.

### Escaleras

Las escaleras adelantan al jugador:

$$
3 \rightarrow 11
$$

$$
15 \rightarrow 19
$$


### Serpientes

Las serpientes retrasan al jugador:

$$
13 \rightarrow 4
$$

$$
17 \rightarrow 9
$$

Estas transiciones modifican el comportamiento probabilístico del sistema, generando aceleraciones y retrocesos durante la partida.


# 3. Modelo Matemático

Sea:

$$
E_i
$$

la esperanza del número de tiradas restantes para terminar el juego cuando el jugador se encuentra en la casilla $i$.

La condición de frontera es:

$$
E_{20}=0
$$

ya que al alcanzar la meta el juego termina inmediatamente.


## 3.1 Ecuación de recurrencia

Desde cualquier casilla existen seis posibles estados futuros.

Por lo tanto:

$$
E_i=
1+
\frac{1}{6}
\sum_{k=1}^{6}
E_{f(i+k)}
$$

donde:

$$
f(x)
$$

representa la función que aplica automáticamente las serpientes y escaleras del tablero.

El término:

$$
1
$$

representa la tirada actual realizada por el jugador.



# 4. Método de Monte Carlo

Para validar el resultado teórico, se realizó una simulación de Monte Carlo.

Cada simulación consiste en:

1. Iniciar en la casilla $0$.
2. Lanzar un dado aleatorio.
3. Aplicar serpientes y escaleras.
4. Contar el número total de movimientos.
5. Repetir el proceso hasta alcanzar la casilla $20$.

Si:

$$
X_i
$$

representa el número de tiradas de la simulación $i$, entonces el valor esperado puede aproximarse mediante:

$$
E[X]
\approx
\frac{1}{N}
\sum_{i=1}^{N} X_i
$$

donde:

$$
N
$$

es el número total de simulaciones.

Según la Ley de los Grandes Números, el promedio muestral converge al valor esperado real conforme:

$$
N \rightarrow \infty
$$



```python
import random

def simular_juego():
    posicion = 0
    turnos = 0
    meta = 20

    especiales = {
        3: 11,   # Escalera
        15: 19,  # Escalera
        13: 4,   # Serpiente
        17: 10   # Serpiente
    }

    while posicion < meta:
        dado = random.randint(1, 6)
        posicion += dado
        turnos += 1

        # Si caemos en escalera o serpiente, nos movemos
        if posicion in especiales:
            posicion = especiales[posicion]

    return turnos

# Ejecutar la simulación n veces
n_simulaciones = 100000
total_turnos = sum(simular_juego() for _ in range(n_simulaciones))

promedio = total_turnos / n_simulaciones

print(f"Resultados tras {n_simulaciones:,} partidas:")
print(f"El número promedio de tiradas es: {promedio:.2f}")
```

# Análisis Estocástico: El Ratón en el Laberinto


# 1. Introducción

En este problema se analiza el comportamiento probabilístico de un ratón dentro de un laberinto representado mediante una red de estados conectados.

El objetivo consiste en determinar la probabilidad de que el ratón alcance la casilla de comida antes de llegar a la casilla de choque eléctrico.

El sistema puede modelarse mediante una cadena de Markov absorbente, ya que existen estados terminales donde el proceso finaliza definitivamente.

# 2. Definición del Problema

El laberinto está formado por los estados:

$$
\{0,1,2,3,4,5,6,7,8\}
$$

donde:

- El estado $7$ representa la comida.
- El estado $8$ representa el choque eléctrico.

El ratón inicia en la casilla:

$$
0
$$

y en cada movimiento selecciona aleatoriamente una de las puertas disponibles en su posición actual.

## 2.1 Estados absorbentes y transitorios

El modelo se clasifica de la siguiente manera.

### Estados absorbentes

Son estados donde el proceso termina:

$$
\{7,8\}
$$

En particular:

$$
P_7 = 1
$$

y

$$
P_8 = 0
$$

---

### Estados transitorios

Son los estados donde el ratón aún puede seguir moviéndose:

$$
\{0,1,2,3,4,5,6\}
$$


# 3. Modelo Probabilístico

Sea:

$$
P_i
$$

la probabilidad de que el ratón llegue a la comida iniciando desde el estado $i$.

Cada transición ocurre con probabilidad uniforme.

Si una casilla tiene $k$ salidas posibles, entonces:

$$
p_{ij}=\frac{1}{k}
$$

para cada vecino accesible.

## 3.1 Ecuaciones de transición

Las probabilidades satisfacen las ecuaciones:

$$
P_i=
\sum_j p_{ij}P_j
$$

lo que genera un sistema lineal asociado a la cadena de Markov.


# 4. Simetría del Laberinto

El diseño del laberinto presenta una estructura aproximadamente simétrica.

Las distancias desde el estado inicial hasta:

- la comida,
- el choque,

son equivalentes en términos de conectividad y número de caminos disponibles.

Debido a esta simetría puede inferirse intuitivamente que:

$$
P(\text{Comida})
=
P(\text{Choque})
=
0.5
$$


# 5. Método de Monte Carlo

Para validar el resultado teórico se implementó una simulación de Monte Carlo.

Cada experimento consiste en:

1. Colocar al ratón en el estado $0$.
2. Elegir movimientos aleatorios válidos.
3. Continuar el recorrido hasta alcanzar:
   la comida ($7$), o
   el choque ($8$).
4. Registrar el resultado final

## 5.1 Aproximación probabilística

Si se realizan "N" simulaciones independientes, la probabilidad de éxito puede aproximarse mediante:

$$
P(E)
\approx
\frac{\text{Ratones que alcanzaron comida}}
{\text{Total de simulaciones}}
$$

Según la Ley de los Grandes Números:

$$
\lim_{N\to\infty}
\frac{\text{Éxitos}}{N}
=
P(E)
$$

# 6. Resultado

Al resolver el sistema de ecuaciones asociado a la cadena de Markov y comparar con la simulación numérica, se obtiene:

$$
\boxed{
P_0=0.5
}
$$

Por lo tanto, la probabilidad de que el ratón alcance la comida iniciando desde la casilla $0$ es:

$$
\boxed{
50\%
}
$$


# 7. Conclusion

La estructura simétrica del laberinto provoca que los dos estados absorbentes tengan la misma probabilidad de ser alcanzados desde el estado inicial.


```python
import random

def simular_raton():
    posicion = 0 # Inicia en la casilla 0
    # Diccionario de conexiones (basado en las puertas de la imagen)
    adyacencia = {
        0: [1, 2],
        1: [0, 3, 7],
        2: [0, 3, 8],
        3: [1, 2, 4, 5],
        4: [3, 6, 7],
        5: [3, 6, 8],
        6: [4, 5],
        7: [7], # Food (absorbente)
        8: [8]  # Shock (absorbente)
    }

    while posicion not in [7, 8]:
        opciones = adyacencia[posicion]
        posicion = random.choice(opciones)

    return 1 if posicion == 7 else 0

# Ejecutar simulación
n_pruebas = 100000
exitos = sum(simular_raton() for _ in range(n_pruebas))

probabilidad = exitos / n_pruebas
print(f"Probabilidad estimada de alcanzar la comida: {probabilidad:.4f}")
```
